```python
import numpy as np

def matmul_strided_manual(A_flat, B_flat, M, N, K, strideA, strideB):
    # Create an output buffer simulating a flat OpenCL __global buffer
    C_flat = np.zeros(M * N, dtype=np.float32)
    
    for i in range(M):
        for j in range(N):
            sum_val = 0.0
            for p in range(K):
                # Manual index calculation using row strides
                idxA = i * strideA + p
                idxB = p * strideB + j
                sum_val += A_flat[idxA] * B_flat[idxB]
            
            idxC = i * N + j
            C_flat[idxC] = sum_val
            
    return C_flat

# --- Example Usage ---
M, N, K = 2, 2, 2
# Set strides larger than width to simulate padding (e.g., for memory alignment)
sA, sB, sC = 4, 4, 4 

# Create flat buffers filled with sample data
A_data = np.array([1, 2, 0, 0, 3, 4, 0, 0], dtype=np.float32) # [1,2] followed by padding
B_data = np.array([5, 6, 0, 0, 7, 8, 0, 0], dtype=np.float32) # [5,6] followed by padding

result_flat = matmul_strided_manual(A_data, B_data, M, N, K, sA, sB)

# Reshape just to verify the results are correct (2x2 part of the buffer)
# print("C[0,0]:", result_flat[0*sC + 0]) # Expected: 1*5 + 2*7 = 19
# print("C[1,1]:", result_flat[1*sC + 1]) # Expected: 3*6 + 4*8 = 50

result_flat
```
